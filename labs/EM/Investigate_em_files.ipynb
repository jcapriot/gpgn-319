```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

# EM31 data

```python
em31 = pd.read_csv('data/labEM3/EM-31_data.csv')
```

```python
em31
```

```python
plt.scatter(em31['Longitude (deg)'], em31['Latitude (deg)'], s=1)
```

```python
plt.plot(em31['time stamp (ms)'], marker='x', linestyle='')

#3857
em31_hcp_per = em31[780:3857]
em31_vcp_per = em31[4049:7015]
em31_hcp_par = em31[7040:]
plt.plot(em31_hcp_per['time stamp (ms)'], marker='x', linestyle='')
plt.plot(em31_vcp_per['time stamp (ms)'], marker='x', linestyle='')
plt.plot(em31_hcp_par['time stamp (ms)'], marker='x', linestyle='')
```

```python
line = em31_hcp_per

plt.figure(figsize=(12, 4), dpi=200)
plt.subplot(1,2,1)
plt.tripcolor(line["Longitude (deg)"], line["Latitude (deg)"], line["In-phase (ppt)"], shading='gouraud')
plt.gca().set_aspect(1)
plt.colorbar()

plt.subplot(1, 2, 2)
plt.tripcolor(line["Longitude (deg)"], line["Latitude (deg)"], line["conductivity (mS/m)"], shading='gouraud')
plt.gca().set_aspect(1)
plt.colorbar()
plt.tight_layout()
```

```python
line = em31_vcp_per

plt.figure(figsize=(12, 4), dpi=200)
plt.subplot(1,2,1)
plt.tripcolor(line["Longitude (deg)"], line["Latitude (deg)"], line["In-phase (ppt)"], shading='gouraud')
plt.gca().set_aspect(1)
plt.colorbar()

plt.subplot(1, 2, 2)
plt.tripcolor(line["Longitude (deg)"], line["Latitude (deg)"], line["conductivity (mS/m)"], shading='gouraud')
plt.gca().set_aspect(1)
plt.colorbar()
plt.tight_layout()
```
