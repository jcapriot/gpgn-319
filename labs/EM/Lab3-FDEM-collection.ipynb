```python
temp = fdem_sens.HalfspaceSensitivity()
info = temp.data_map
```

```python
def plot_semilogx_scaled(x, y, **kwargs):
    abs_max = np.abs(y).max()
    y = y/abs_max
    plt.semilogx(x, y, **kwargs)
```

```python
plot_semilogx_scaled(temp.depths, temp.sens[info['em31']['vcp']['waist']['imag']])
plot_semilogx_scaled(temp.depths, temp.sens[info['em31']['hcp']['waist']['imag']])
```

```python
plot_semilogx_scaled(temp.depths, temp.sens[info['em31']['hcp']['surface']['imag']])
plot_semilogx_scaled(temp.depths, temp.sens[info['em31']['hcp']['waist']['imag']])
```

```python
plot_semilogx_scaled(temp.depths, temp.sens[info['em34']['hcp'][10]['imag']])
plot_semilogx_scaled(temp.depths, temp.sens[info['em34']['vcp'][10]['imag']])
plot_semilogx_scaled(temp.depths, temp.sens[info['em34']['vca'][10]['imag']])
```

```python
plot_semilogx_scaled(temp.depths, temp.sens[info['em34']['hcp'][10]['imag']])
plot_semilogx_scaled(temp.depths, temp.sens[info['em34']['hcp'][20]['imag']])
plot_semilogx_scaled(temp.depths, temp.sens[info['em34']['hcp'][40]['imag']])
```
