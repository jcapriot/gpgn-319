Next we are going to investigate the sensitivity of the FDEM equipment we have used. You should be able to then visualize how the response changes with different survey configurations, and how you could use multiple configurations to sample to earth with depth.

Let's represent the forward operation of collecting a single data point as a function of our depth dependent conductivity $\sigma(z)$ as:
$$
d = f(\sigma(z))
$$

What we want to investigate is how a change in conductivity at any point in depth, changes the data

$$
\frac{\partial d}{\partial \sigma}\bigg|_z = \frac{\partial f}{\partial \sigma}\bigg|_z
$$


You should have this configuration of survey parameters to play with, in addition to the frequency and orientation of the transmitter and receiver
