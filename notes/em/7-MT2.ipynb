## MT Assumptions:
What were they?

* The source is very far such that the waves are approximately plane waves.

* In the quasistatic regime
    * EM velocity in the air is much greater than velocity in the Earth
    * All waves travel downwards into the earth $\vec{k} || \hat{z}$

Which direction are the $E$ and $B$ fields polarized in the Earth?

### 1D MT

Assume plane wave traveling in the -z-direction (and we've arbitrarily rotated our axes about $\hat{z}$
to have $\vec E_0 || \hat{x}$:

$$
\vec E(z) = E_0 e^{i(-k_z z - \omega t)}\hat{x}
$$

The B field is (from Faraday's Law)

\begin{align}
\nabla \times \vec{E} &= - \frac{\partial \vec{B}}{\partial t}\\
i \vec{k} \times \vec{E} &= i\omega \vec{B}\\
\vec{k} \times \vec{E} &= \omega \vec{B}\\
-k_z E_0 e^{i(k_z z - \omega t)} \hat{z} \times \hat{x} &= \omega \vec{B}\\
\frac{-k_z E_0}{\omega} e^{i(k_z z - \omega t)} \hat{y} &= \vec{B}
\end{align}

or:
$$
\vec{H} = \frac{-k_z E_0}{\omega\mu} e^{i(k_z z - \omega t)} \hat{y}
$$

Looking at the ratio of $E_x$ and $H_y$ allows us to cancel out the magntidue of the source!

\begin{align}
\frac{E_x}{H_y} = \frac{-\omega \mu}{k_z}
\end{align}

and $k_z$ is a function of $\omega$, $\sigma$, and $\mu$ (and $\epsilon$)

$$
k = \sqrt{\mu \varepsilon \omega^2 - i \omega\mu\sigma}
$$

And therefore $\frac{E_x}{H_y}$ is a measurent of the physical properties, independant of the source strength.

### In a halfspace
in the quasi-static regime, this yields:

\begin{align}
\frac{E_x}{H_y} = \sqrt{\frac{\mu\omega}{\sigma}} 
\end{align}

which we re-arrange to give us an apparent conductivity/resistivity transformation for Magnetotellurics

## EM wavemodes
All electromagnetic waves can be broken down in to two pieces called "modes".

* Transverse Electric (TE) mode.
* Transverse Magnetic (TM) mode.

### TE
A TE mode wvae only has a component of $\vec{E}$ perpendicular to some axis (usually, but not always, it's propogation direction), and a component of $\vec{B}$ also perpendicular to that axis.

## 2D MT
* Assume all derivatives w.r.t. $y$ are zero.

Break it into a TE mode with $\vec{E}$ perpendicular to strike ($\hat{y}$), $E_y = 0$, $E_x \in R$ and $E_z \in R$

and a TM mode with $\vec{H}$ perpendicular to strike.

The TE mode implies:

\begin{align}
\nabla \times \vec{E} = i \omega \vec{H}\\
\begin{pmatrix}
\partial x\\
0\\
\partial z\\
\end{pmatrix}\times \vec{E} = i \omega \vec{H}\\
(\partial_z E_x - \partial_x E_z )\hat{y} = i \omega \vec{H}
\end{align}

So:
* $\vec{H} || \hat{y}$!
* At surface $E_z = 0$ (no current flows into the air)
* Still have $E_x = Z_{xy}H_y$
* $H_x$ for TE mode is 0

### TM mode
* By a similar argument, and using Ampere's Law: $\nabla \times \vec{H} = \sigma \vec{E}$

If $\vec{H} \perp \hat{y}$, then $H_y = 0$ and there are no assumptions on $H_x$ and $H_z$ (except that $H_z=0$ at the surface):

Yields only a $E_y$ component of the electric field.

### So what?
This means that in 2D, the relationship between $\vec{E}$ and $\vec{H}$ at the surface must be:

\begin{align}
\begin{pmatrix}
E_x \\
E_y \\
\end{pmatrix}
\begin{pmatrix}
0 & Z_{xy} \\
Z_{yx} & 0 \\
\end{pmatrix}
\begin{pmatrix}
H_x \\ H_y
\end{pmatrix}
\end{align}

* Basically the same as 1D, except now $Z_{xy} \neq -Z_{yx}$ (in general)
* $Z_{xx}$ and $Z_{yy}$ only zero if $x$ and $y$ are aligned with strike!

### What about 3D?
No more assumptions on derivatives...

Still split into TE and TM modes, but now TE mode gives us,

$$
E_x = Z_{xx} H_x + Z_{xy} H_y
