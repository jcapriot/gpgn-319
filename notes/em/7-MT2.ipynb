## The MT Experiment:

Monitor Electric and Magnetic fields on the surface of the Earth

Compaire the ratio of the fields' amplitudes and the difference of their phases to infer electrical properties.

## MT Assumptions:
What were they?

* The source is very far such that the waves are approximately plane waves.

* In the quasistatic regime
    * EM velocity in the air is much greater than velocity in the Earth
    * All waves travel downwards into the earth $\vec{k} || \hat{z}$

Which direction are the $E$ and $H$ fields polarized in the Earth?

The general MT impedance tensor

$$
\begin{pmatrix}
E_x(\omega) \\ 
E_y(\omega)
\end{pmatrix} = 
\begin{pmatrix}
Z_{xx}(\omega) & Z_{xy}(\omega) \\ 
Z_{yx}(\omega) & Z_{yy}(\omega)
\end{pmatrix}
\begin{pmatrix}
H_x(\omega) \\ 
H_y(\omega)
\end{pmatrix}
$$

$Z(\omega)$ is the transfer function between the horizontal components of $\vec{E}$ and $\vec{H}$

### 1D MT

Assume plane wave traveling in the -z-direction (and we've arbitrarily rotated our axes about $\hat{z}$
to have $\vec E_0 || \hat{x}$:

$$
\vec E(z) = E_0 e^{i(\omega t +k_z z)}\hat{x}
$$

The B field is (from Faraday's Law)

\begin{align}
\nabla \times \vec{E} &= - \frac{\partial \vec{B}}{\partial t}\\
-i \vec{k} \times \vec{E} &= -i\omega \vec{B}\\
\vec{k} \times \vec{E} &= \omega \vec{B}\\
-k_z E_0 e^{i(\omega t +k_z z)} \hat{z} \times \hat{x} &= \omega \vec{B}\\
\frac{-k_z E_0}{\omega} e^{i(\omega t +k_z z)} \hat{y} &= \vec{B}
\end{align}

Using the constitutive relation: $\frac{1}{\mu}\vec{B} = \vec{H}$

$$
\vec{H} = \frac{-k_z E_0}{\omega \mu} e^{i(\omega t +k_z z)} \hat{y}
$$

Looking at the ratio of $E_x$ and $H_y$ allows us to cancel out the magnitude of the source!

\begin{align}
\frac{E_x}{H_y} = \frac{-\mu\omega}{k_z}
\end{align}

and $k_z$ is a function of $\omega$, $\sigma$, and $\mu$ (and $\epsilon$)

$$
k_z = \sqrt{\mu \varepsilon \omega^2 - i \omega\mu\sigma}
$$

And therefore $\frac{E_x}{H_y}$ is a measurent of the physical properties, independant of the source strength.

### 1D Impedance Tensor

$$
\begin{pmatrix}
E_x(\omega) \\ 
E_y(\omega)
\end{pmatrix} = 
\begin{pmatrix}
0 & Z_{xy}(\omega) \\ 
Z_{yx}(\omega) &0
\end{pmatrix}
\begin{pmatrix}
B_x(\omega) \\ 
B_y(\omega)
\end{pmatrix}
$$
with
$$Z_{xy} = - Z_{yx}$$

In 1D:
 * The diagonal components are both zero
 * The off diagonals are negatives of each other

### In a quasi-static halfspace
in the quasi-static regime ( $k \approx \sqrt{i \omega \mu \sigma}$), this yields:

$$k \approx \sqrt{i \omega \mu \sigma} = \sqrt{\frac{\omega \mu \sigma}{2}}(1 + 1i) $$

\begin{align}
\frac{E_x}{H_y} &=  -\frac{\omega \mu}{\sqrt{\frac{\omega \mu \sigma}{2}}(1 + 1i)}\\
&= -\frac{\omega  \mu\sqrt{\frac{2}{\omega\mu\sigma}}}{(1 + 1i)}\\
&= \sqrt{\frac{2\omega  \mu}{\sigma}}\frac{-1}{(1 + 1i)}\\
&= \sqrt{\frac{2\omega  \mu}{\sigma}}\frac{-1}{\sqrt{2}e^{i\pi/4}}\\
&= \sqrt{\frac{\omega  \mu}{\sigma}}(-e^{i (-\pi/4)})\\
&= \sqrt{\frac{\omega  \mu}{\sigma}}(-e^{i (-\pi/4)})\\
&= \sqrt{\frac{\omega  \mu}{\sigma}}e^{i (\pi/4)} = \sqrt{\frac{\omega  \mu}{\sigma}}(1 + 1i)
\end{align}

The impedance component is a complex number:

$$Z_{xy} = \frac{E_x}{H_y}$$

who's amplitude is $\sqrt{\frac{\omega\mu}{\sigma}}$, and phase is $45^o$.

We re-arrange the amplitude to give us an apparent conductivity/resistivity transformation for Magnetotellurics

$$
\sigma_{a,xy} = \omega \mu \left|\frac{H_x}{E_y}\right|^2 = \omega \mu |Z_{xy}|^{-2}
$$
or
$$
\rho_{a,xy} = \frac{1}{\omega \mu} |Z_{xy}|^2
$$
* The complex amplitude squared of the impedance is proportional to resistivity

### The units of Z

* $\vec{E}$ is in $\frac{V}{m}$
* $\vec{H}$ is in $\frac{A}{m}$
* $Z$ must then be $V/A$ which is...

* Z is in units of $\Omega$
* It is an [impedance](https://en.wikipedia.org/wiki/Electrical_impedance) after all!

## EM wavemodes
All electromagnetic waves can be broken down in to two pieces called "modes".

* Transverse Electric (TE) mode.
* Transverse Magnetic (TM) mode.

### TE
A TE mode wave only has a component of $\vec{E}$ perpendicular to some axis (usually it's propogation direction)

### TM
And the corresponding mode such the $\vec{H}$ only has a field in this same plane.

## 2D MT
* Rotate axes s.t. $\hat{y}$ is the strike direction
* Assume all derivatives w.r.t. $y$ are zero.

Break it into a TE mode (perpendicular to $\hat{z}$), and with $\vec{E}$ parallel to strike ($\vec{E}||\hat{y}$):

The TE mode implies:

\begin{align}
\nabla \times \vec{E} = -i \omega \vec{B}\\
\begin{pmatrix}
\partial x\\
0\\
\partial z\\
\end{pmatrix}\times \vec{E} = - i \omega \vec{B}\\
-\partial_z E_y \hat{x} + \partial_x E_y \hat{z} = - i \mu\omega \vec{H}
\end{align}

So:
* TE mode implies $H_y=0$
* This TE mode gives us an equation for $E_y$ then in terms of $H_x$ (or $H_z$)
$$ \partial_z E_y = i\mu\omega H_x $$

### TM mode
If $\vec{B} || \hat{y}$ then:

* By a similar argument, and using Ampere's Law:

\begin{align}
\nabla \times \vec{H} &= \sigma \vec{E}\\
\begin{pmatrix}
\partial x\\
0\\
\partial z\\
\end{pmatrix}\times \vec{H} &= \sigma \vec{E}\\
-\partial_z H_y \hat{x} + \partial_x H_y \hat{z} = \sigma \vec{E}
\end{align}

Implies $E_x$ and $E_z$ are related to $H_y$, and we get an equation for $E_x$ in terms of $H_y$
$$
-\partial_z H_y = \sigma E_x
$$

### So what?
This means that in 2D, the relationship between $\vec{E}$ and $\vec{H}$ at the surface must be:

\begin{align}
\begin{pmatrix}
E_x \\
E_y \\
\end{pmatrix}
\begin{pmatrix}
0 & Z_{xy} \\
Z_{yx} & 0 \\
\end{pmatrix}
\begin{pmatrix}
H_x \\ H_y
\end{pmatrix}
\end{align}

* There is no relationship between $E_y$ and $H_y$, (or $E_x$ and $H_x$)
* Basically the same as 1D, except now $Z_{xy} \neq -Z_{yx}$ (in general)

> $Z_{xx}$ and $Z_{yy}$ only zero if $x$ and $y$ are aligned with strike!

$$
\begin{vmatrix}
\hat{x} & \hat{y} & \hat{z} \\ 
\partial_x & \partial_y & \partial_z \\
f_x & f_y & f_z
\end{vmatrix} =
(\partial_y f_z - \partial_z f_y)\hat{x} + (\partial_z f_x - \partial_x f_z)\hat{y} + (\partial_x f_y - \partial_y f_x)\hat{z}
$$

### What about 3D?
No more assumptions on derivatives ($\partial_y \neq 0$ in general)...

And, can no longer split into TE and TM modes, due to the conductivity structure arbitrarily re-orienting the fields.

### From Faraday's Law:

\begin{align}
\nabla \times \vec{E} &= -i \frac{\omega}{\mu} \vec{H}\\
(\partial_y E_z - \partial_z E_y)\hat{x} + (\partial_z E_x - \partial_x E_z)\hat{y} + (\partial_x E_y - \partial_y E_x)\hat{z} &= -i \omega\mu \vec{H}
\end{align}

Gives three sets of equations

1. $\partial_y E_z - \partial_z E_y = -i\omega \mu H_x$
2. $\partial_z E_x - \partial_x E_z = -i\omega \mu H_y$
3. $\partial_x E_y - \partial_y E_x = -i\omega \mu H_z$

### From Ampere's Law:

\begin{align}
\nabla \times \vec{H} = \sigma\vec{E}\\
(\partial_y H_z - \partial_z H_y)\hat{x} + (\partial_z H_x - \partial_x H_z)\hat{y} + (\partial_x H_y - \partial_y H_x)\hat{z} &= \sigma\vec{E}
\end{align}



Gives three sets of equations

4. $\partial_y H_z - \partial_z H_y = -i\sigma E_x$
5. $\partial_z H_x - \partial_x H_z = -i\sigma E_y$
6. $\partial_x H_y - \partial_y H_x = -i\sigma E_z$

### So..
4\. says $E_x$ is dependant upon $H_y$ and $H_z$

3\. says $H_z$ is dependant upon $E_y$ and $E_x$

5\. says $E_y$ is dependant upon $H_x$ and $H_z$

Thus through a bit of a roundabout way,
$E_x$ is dependant upon $H_x$

Similarly $E_y$ is depedant upon $H_y$

So we get a full tensor relationship between the horizontal components of $\vec{E}$ and $\vec{H}$ when there are 3D structures:

$$
\begin{pmatrix}
E_x(\omega) \\ 
E_y(\omega)
\end{pmatrix} = 
\begin{pmatrix}
Z_{xx}(\omega) & Z_{xy}(\omega) \\ 
Z_{yx}(\omega) & Z_{yy}(\omega)
\end{pmatrix}
\begin{pmatrix}
H_x(\omega) \\ 
H_y(\omega)
\end{pmatrix}
$$

## Calculating the impedance tensor

How do we get these quantities in general?
For the impedance relationship above we have two equations, each with two unknowns, clearly this is underdetermined...

1. $E_x = Z_{xx} H_x + Z_{xy} H_y$
2. $E_y = Z_{xy} H_x + Z_{yy} H_y$

There's an infinite number of valid values for $Z_{xx}, Z_{xy}, Z_{yx}, Z_{yy}$ that would satisfy the above equation

### Idea:
We have a long time series of data, why don't we split it in half, and get two values for each field component?


Looking at the first equation then would give us:

\begin{align}
E_{x,1} &= Z_{xx} H_{x,1} + Z_{xy} H_{y,1} \\
E_{x,2} &= Z_{xx} H_{x,2} + Z_{xy} H_{y,2} \\
\end{align}

Now we've got two equations with two unknowns, we could solve for $Z_{xx}$ and $Z_{xy}$!

### Why stop at 2?
Let's split it up $N$ times!

\begin{align}
E_{x,1} &= Z_{xx} H_{x,1} + Z_{xy} H_{y,1} \\
E_{x,2} &= Z_{xx} H_{x,2} + Z_{xy} H_{y,2} \\
&\vdots\\
E_{x,N} &= Z_{xx} H_{x,N} + Z_{xy} H_{y,N} \\
\end{align}

And then solve by a Least Squares approach (but remember everything is *complex*)

Let's measure the difference using $v = E_x - Z_{xx} H_x - Z_{xy} H_{y}$, and measure its [*Frobenius*](https://en.wikipedia.org/wiki/Frobenius_inner_product) norm.

$$
\phi = \sum_{i=1}^{N} <v_i, v_i> = \sum_{i=1}^{N} v_iv_i^*=\sum_{i=1}^N (E_{x,i} - Z_{xx}H_{x,i} - Z_{xy}H_{y,i})(E_{x,i} - Z_{xx}H_{x,i} - Z_{xy}H_{y,i})^*
$$

or in terms of a matrix-vector equation:

$$\phi = ||AZ_x - b||_F^2 = (AZ_{x} - b)(AZ_{x} - b)^*
$$

with
$$
A = \begin{pmatrix}
H_{x, 1} & H_{y,1}\\
H_{x, 2} & H_{y,2}\\
\vdots\\
H_{x, N} & H_{y,N}\\
\end{pmatrix},
Z_x = \begin{pmatrix}
Z_{xx}\\
Z_{xy}
\end{pmatrix},
b = \begin{pmatrix}
E_{x,1}\\
E_{x,2}\\
\vdots\\
E_{x,N}
\end{pmatrix}
$$

### Complex least squares:
To minimize, take a [derivative](https://en.wikipedia.org/wiki/Wirtinger_derivatives) and set it equal to zero:
> special care must be taken because $Z_{ij}$ is a complex number, and derivatives with respect to complex numbers are not *quite* the same as derivatives with respect to real numbers

$$ d\phi = (A Z_x - b)^*: A dZ_x$$

$$ \frac{d\phi}{dZ_x} = A^T (A Z_x - b)^* =>0 $$
$$ A^T b^*=A^T (A Z_x)^*$$
$$  A^H b = A^H A Z_x$$

> $ A^H $ means $A$ conjugate tranpose

Looking at each row of the above equation, the first row gives:
$$
\sum E_{x,i} H_{x_i}^* =Z_{xx}\sum H_{x,i}H_{x,i}^*  + Z_{xy}\sum H_{y,i}H_{x,i}^*
$$

and the second row:
$$
\sum E_{x,i} H_{y_i}^* = Z_{xx}\sum H_{x,i}H_{y,i}^*  + Z_{xy}\sum H_{y,i}H_{y,i}^*
$$


### What is $\sum H_{x,i}(\omega) H_{x,i}(\omega)$ doing?
$H_x$ is the frequency transformed $h_x$

* All of our $H_{x,i}$ are just transformed pieces of $h_x$

* Multiplication in frequency domain is convolution in the time domain

* Conjugate operator in frequency domain is a time-reversal in the time domain

* $H_{x,i}H_{x,i}^*$ Implies we are just convolving $h_{x,i}$ with itself reversed

* Break into an arbitrary number of items, then:

$$
\sum H_{x,i}(\omega) H_{x,i}^*(\omega) = H_x(\omega)H_x^*(\omega)
$$

* $\sum H_{x,i}(\omega) H_{x,i}^*(\omega)$ The [*auto-correlation*](https://en.wikipedia.org/wiki/Autocorrelation) as a function of frequency of $h_x$

* $\sum H_{x,i}(\omega) H_{y,i}^*(\omega)$ The [*cross-correlation*](https://en.wikipedia.org/wiki/Cross-correlation) as a function of frequency of $h_x$ and $h_y$

* let $<UV^*>$ represent the power spectrum of the cross-correlation $(u \star v)$ (auto-correlation if they're the same) of the two time series $u$ and $v$.

Re-writing in terms of cross/auto power-spectrums:

$$<E_xH_x^*> = Z_{xx} <H_xH_x^*> + Z_{xy} <H_yH_x^*>$$
$$<E_xH_y^*> = Z_{xx} <H_xH_y^*> + Z_{xy} <H_yH_y^*>$$

Solve for $Z_{xx}$ and $Z_{xy}$:

$$Z_{xx} = \frac{<H_yH_y^*><E_xH_x^*> - <H_yH_x*><E_xH_y^*>}{<H_xH_x^*> <H_yH_y^*> - <H_xH_y^*><H_yH_x^*>}$$
$$Z_{xy} = \frac{<H_xH_x^*><E_xH_y^*> - <H_xH_y*><E_xH_x^*>}{<H_xH_x^*> <H_yH_y^*> - <H_xH_y^*><H_yH_x^*>}$$

### Similarly
Using the second row of the impedance tensor:

$$<E_yH_x^*> = Z_{yx} <H_xH_x^*> + Z_{yy} <H_yH_x^*>$$
$$<E_yH_y^*> = Z_{yx} <H_xH_y^*> + Z_{yy} <H_yH_y^*>$$


$$Z_{yx} = \frac{<H_yH_y^*><E_yH_x^*> - <H_yH_x*><E_yH_y^*>}{<H_xH_x^*> <H_yH_y^*> - <H_xH_y^*><H_yH_x^*>}$$
$$Z_{yy} = \frac{<H_xH_x^*><E_yH_y^*> - <H_xH_y*><E_yH_x^*>}{<H_xH_x^*> <H_yH_y^*> - <H_xH_y^*><H_yH_x^*>}$$

### A note on cross-correlations:

Cross-correlations are a measure of similarity between two signals as a function of the displacement of one relative to the other.


If $\tilde{u}$ has noise, and $\tilde{v}$ also has noise, $(\tilde{u} \star \tilde{v})$ will estimate $(u \star v )$ as:
* unbiased if the noise in each measurements are uncorrelated.
* upward biased if the noise is correlated
* downward biased if the noise is anti-correlated

* Auto-correlation $(\tilde{u} \star \tilde{u})$ will be an upward biased estimate of $(u \star u)$.
