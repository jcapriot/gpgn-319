## The MT Experiment:

Monitor Electric and Magnetic fields on the surface of the Earth

Compaire the ratio of the fields' amplitudes and the difference of their phases to infer electrical properties.

## MT Assumptions:
What were they?

* The source is very far such that the waves are approximately plane waves.

* In the quasistatic regime
    * EM velocity in the air is much greater than velocity in the Earth
    * All waves travel downwards into the earth $\vec{k} || \hat{z}$

Which direction are the $E$ and $H$ fields polarized in the Earth?

The general MT impedance tensor

$$
\begin{pmatrix}
E_x(\omega) \\ 
E_y(\omega)
\end{pmatrix} = 
\begin{pmatrix}
Z_{xx}(\omega) & Z_{xy}(\omega) \\ 
Z_{yx}(\omega) & Z_{yy}(\omega)
\end{pmatrix}
\begin{pmatrix}
H_x(\omega) \\ 
H_y(\omega)
\end{pmatrix}
$$

$Z(\omega)$ is the transfer function between the horizontal components of $\vec{E}$ and $\vec{H}$

### 1D MT

Assume plane wave traveling in the -z-direction (and we've arbitrarily rotated our axes about $\hat{z}$
to have $\vec E_0 || \hat{x}$:

$$
\vec E(z) = E_0 e^{i(\omega t +k_z z)}\hat{x}
$$

The B field is (from Faraday's Law)

\begin{align}
\nabla \times \vec{E} &= - \frac{\partial \vec{B}}{\partial t}\\
-i \vec{k} \times \vec{E} &= -i\omega \vec{B}\\
\vec{k} \times \vec{E} &= \omega \vec{B}\\
-k_z E_0 e^{i(\omega t +k_z z)} \hat{z} \times \hat{x} &= \omega \vec{B}\\
\frac{-k_z E_0}{\omega} e^{i(\omega t +k_z z)} \hat{y} &= \vec{B}
\end{align}

Using the constitutive relation: $\frac{1}{\mu}\vec{B} = \vec{H}$

$$
\vec{H} = \frac{-k_z E_0}{\omega \mu} e^{i(\omega t +k_z z)} \hat{y}
$$

Looking at the ratio of $E_x$ and $H_y$ allows us to cancel out the magnitude of the source!

\begin{align}
\frac{E_x}{H_y} = \frac{-\mu\omega}{k_z}
\end{align}

and $k_z$ is a function of $\omega$, $\sigma$, and $\mu$ (and $\epsilon$)

$$
k_z = \sqrt{\mu \varepsilon \omega^2 - i \omega\mu\sigma}
$$

And therefore $\frac{E_x}{H_y}$ is a measurent of the physical properties, independant of the source strength.

### 1D Impedance Tensor

$$
\begin{pmatrix}
E_x(\omega) \\ 
E_y(\omega)
\end{pmatrix} = 
\begin{pmatrix}
0 & Z_{xy}(\omega) \\ 
Z_{yx}(\omega) &0
\end{pmatrix}
\begin{pmatrix}
B_x(\omega) \\ 
B_y(\omega)
\end{pmatrix}
$$
with
$$Z_{xy} = - Z_{yx}$$

In 1D:
 * The diagonal components are both zero
 * The off diagonals are negatives of each other

### In a quasi-static halfspace
in the quasi-static regime ( $k \approx \sqrt{i \omega \mu \sigma}$), this yields:

$$k \approx \sqrt{i \omega \mu \sigma} = \sqrt{\frac{\omega \mu \sigma}{2}}(1 + 1i) $$

\begin{align}
\frac{E_x}{H_y} &=  -\frac{\omega \mu}{\sqrt{\frac{\omega \mu \sigma}{2}}(1 + 1i)}\\
&= -\frac{\omega  \mu\sqrt{\frac{2}{\omega\mu\sigma}}}{(1 + 1i)}\\
&= \sqrt{\frac{2\omega  \mu}{\sigma}}\frac{-1}{(1 + 1i)}\\
&= \sqrt{\frac{2\omega  \mu}{\sigma}}\frac{-1}{\sqrt{2}e^{i\pi/4}}\\
&= \sqrt{\frac{\omega  \mu}{\sigma}}(-e^{i (-\pi/4)})\\
&= \sqrt{\frac{\omega  \mu}{\sigma}}(-e^{i (-\pi/4)})\\
&= \sqrt{\frac{\omega  \mu}{\sigma}}e^{i (\pi/4)} = \sqrt{\frac{\omega  \mu}{\sigma}}(1 + 1i)
\end{align}

The impedance component is a complex number then:

$$Z_{xy} = \frac{E_x}{H_y}$$

who's amplitude is $\sqrt{\frac{\omega\mu}{\sigma}}$, and phase is $45^o$.

We re-arrange the amplitude to give us an apparent conductivity/resistivity transformation for Magnetotellurics

$$
\sigma_{a,xy} = \omega \mu \left|\frac{H_x}{E_y}\right|^2 = \omega \mu |Z_{xy}|^{-2}
$$
or
$$
\rho_{a,xy} = \frac{1}{\omega \mu} |Z_{xy}|^2
$$
* The complex amplitude squared of the impedance is proportional to resistivity

### The units of Z

* $\vec{E}$ is in $\frac{V}{m}$
* $\vec{H}$ is in $\frac{A}{m}$
* $Z$ must then be $V/A$ which is...

* Z is in units of $\Omega$
* It is an [impedance](https://en.wikipedia.org/wiki/Electrical_impedance) after all!

## EM wavemodes
All electromagnetic waves can be broken down in to two pieces called "modes".

* Transverse Electric (TE) mode.
* Transverse Magnetic (TM) mode.

### TE
A TE mode wave only has a component of $\vec{E}$ perpendicular to some axis (usually it's propogation direction)

### TM
And the corresponding mode such the $\vec{H}$ only has a field in this same plane.

## 2D MT
* Assume all derivatives w.r.t. $y$ are zero.
* Vertically incident planewave ($E_z = 0$ and $H_z=0$ at the surface)

Break it into a TE mode (perpendicular to $\hat{z}$ rotated s.t. $\vec{E}$ parallel to strike ($\hat{y}$), $\vec{E} || \hat{y}$

The TE mode implies:

\begin{align}
\nabla \times \vec{E} = -i \omega \vec{B}\\
\begin{pmatrix}
\partial x\\
0\\
\partial z\\
\end{pmatrix}\times \vec{E} = - i \omega \vec{B}\\
-\partial_z E_y \hat{x} + \partial_x E_y \hat{z} = - i \frac{\omega}{\mu} \vec{H}
\end{align}

So:
* TE mode implies $E_x=0$, $E_z=0$, $B_y=0$
* This TE mode gives us an equation for $E_y$ then in terms of $H_x$ (or $H_z$)
$$ \partial_z E_y = i\frac{\omega}{\mu}H_x $$

### TM mode
If $\vec{B} || \hat{y}$ then:

* By a similar argument, and using Ampere's Law:

\begin{align}
\nabla \times \vec{H} &= \sigma \vec{E}\\
\begin{pmatrix}
\partial x\\
0\\
\partial z\\
\end{pmatrix}\times \vec{H} &= \sigma \vec{E}\\
-\partial_z H_y \hat{x} + \partial_x H_y \hat{z} = \sigma \vec{E}
\end{align}

Implies $E_x$ and $E_z$ are related to $H_y$, and we get an equation for $E_x$ in terms of $H_y$
$$
-\partial_z H_y = \sigma E_x
$$

### So what?
This means that in 2D, the relationship between $\vec{E}$ and $\vec{H}$ at the surface must be:

\begin{align}
\begin{pmatrix}
E_x \\
E_y \\
\end{pmatrix}
\begin{pmatrix}
0 & Z_{xy} \\
Z_{yx} & 0 \\
\end{pmatrix}
\begin{pmatrix}
H_x \\ H_y
\end{pmatrix}
\end{align}

* There is no relationship between $E_y$ and $H_y$, (or $E_x$ and $H_x$)
* Basically the same as 1D, except now $Z_{xy} \neq -Z_{yx}$ (in general)

> $Z_{xx}$ and $Z_{yy}$ only zero if $x$ and $y$ are aligned with strike!

### What about 3D?
No more assumptions on derivatives ($\partial_y \neq 0$ in general)...

Can still split into TE and TM modes:

#### 3D TE:
$\vec{E}\perp\vec{z}$

\begin{align}
\nabla \times \vec{E} &= -i \frac{\omega}{\mu} \vec{H}\\
-\partial_z E_y \hat{x} + \partial_z E_x \hat{y} + (\partial_x E_y - \partial_y E_x)\hat{z} &= -i \frac{\omega}{\mu} \vec{H}
\end{align}

So the TE mode gives us a relationship between $E_x$ and $H_y$ and $E_y$ and $H_x$:

* $\partial_z E_y = i \frac{\omega}{\mu} H_x$
* $\partial_z E_x = -i \frac{\omega}{\mu} H_y$

#### 3D TM mode
$\vec{H}\perp\vec{z}$

\begin{align}
\nabla \times \vec{H} = \sigma\vec{E}\\
-\partial_z H_y \hat{x} + \partial_z H_x \hat{y} + (\partial_x H_y - \partial_y H_x)\hat{z} &= \sigma\vec{E}
\end{align}

Then:
* $-\partial_z H_y = \sigma E_x$
* $\partial_z H_x = \sigma E_y$

$$
E_y = Z_{yx} H_x + Z_{yy} H_y
$$

and the TM mode gives us that $E_x$ is also dependant on $H_x$ and $H_y$:

$$
E_x = Z_{xx} H_x + Z_{yx} H_y
$$

So we get a full tensor relationship between the horizontal components of $\vec{E}$ and $\vec{H}$:

$$
\begin{pmatrix}
E_x(\omega) \\ 
E_y(\omega)
\end{pmatrix} = 
\begin{pmatrix}
Z_{xx}(\omega) & Z_{xy}(\omega) \\ 
Z_{yx}(\omega) & Z_{yy}(\omega)
\end{pmatrix}
\begin{pmatrix}
H_x(\omega) \\ 
H_y(\omega)
\end{pmatrix}
$$
