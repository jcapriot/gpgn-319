### Basic Assumptions
* The "source" is very far away
* Such that the incoming EM waves are planar

In the air, where conductivity is $\sigma \approx 0$ the wave travels with a velocity of:

$$v_{air} = \frac{1}{\sqrt{\mu\varepsilon}} $$

In the earth where conductivity is larger, we have:
$$v_{earth} = \sqrt{\frac{2\omega}{\mu \sigma }}$$

As a wave hits the interface (with some angle $\theta_{air}$) it refracts into the earth with an angle according to Snell's law ($\theta_{earth}$).

$$
\frac{\sin\theta_{air}}{\sin\theta_{earth}} = \frac{v_{air}}{v_{earth}}
$$

The ratio of

\begin{align}
\frac{v_{air}}{v_{earth}} &= \sqrt{\frac{\sigma}{2\omega\varepsilon}}
\end{align}

Which will go to $\infty$ as either $\sigma \rightarrow \infty$ or $\omega \rightarrow 0$

In our quasistatic regime $\sigma >> \omega\varepsilon$, which implies:

$$
\frac{\sin\theta_{air}}{\sin\theta_{earth}} = \infty
$$
$$
\sin\theta_{earth} = 0
$$
$$
\theta_{earth} = 0
$$

So in the quasistatic regime
* The incident angle of the planewave doesn't matter, it always comes in and travels straight into the Earth

### The MT experiment

* Monitor the Electric and Magnetic fields at the surface.
* Turns out, the transfer function between the two of them is independant of the source.

$$ \vec{E} = Z \vec{H} $$
$$Z = \begin{pmatrix}
Z_{xx} & Z_{xy}\\
Z_{yx} & Z_{yy}
\end{pmatrix}
$$

* $Z$ is dependant on the physical properties!
