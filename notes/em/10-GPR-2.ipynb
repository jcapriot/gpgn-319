## A Review
* GPR operates in the wave-regime of electromagnetics
* Waves travel through materials with a velocity depending on it's physical properties
$$v_{wave} \approx \sqrt{\frac{1}{\mu \varepsilon}}=\frac{c}{\sqrt{\varepsilon_r}}, \hspace{20pt} d \approx 
\frac{2}{\sigma}\sqrt{\frac{\varepsilon}{\mu}}$$
* Measures very rapidly (nano-second time scale)
* They refract based on velocities.
* Materials will always be slower than a vacuum ($\varepsilon_r \geq 1$, $v \leq c$)

### Dielectric Permittivity

* Dielectric permitivity is usually the key physical property

<div class=row>
<div class=column>

The tendancy of a material's electrical dipoles to align (and be displaced) in an electrical field.

* Constitutive relationship between electric field ($\vec{E}$), electrical polarization  ($\vec{P}$) and dielectric displacement ($\vec{D}$):
$$
\vec{D} = \varepsilon \vec{E} + \vec{P}
$$

  > (Think $\frac{1}{\mu}\vec{B} = \vec{H} + \vec{M}$)
* Has units of $F/m$ (Farads per meter)

* Water content is biggest control of $\varepsilon$ that we encounter
* Conductivity controls how deep we see

### At boundaries:
<div class=row>
<div class=column>

* EM waves will reflect (and transmit) at boundaries of different electrical impedance $\eta = \sqrt{\frac{\mu}{\varepsilon}}$

  The *Electric* field is reflected and (possibly) transmitted:
  \begin{align}
  E_{0,R} = \frac{\eta_2 - \eta_1}{\eta_2 + \eta_1} E_0\\
  E_{0,T} = \frac{2 \eta_2}{\eta_2 + \eta_1} E_0
  \end{align}

* EM waves will refract at boundaries according to Snell's Law:
  $$ \frac{\sin\theta_1}{V_1} = \frac{\sin\theta_2}{V_2}$$
</div>
<div class=column>

![refraction](https://gpg.geosci.xyz/_images/GPR_refraction.png)
</div>
</div>

### Tx Waveforms:
A GPR antenna uses a short duration oscillating current, a.k.a a **wavelet**, as its waveform.

* The reported frequency for GPR usually refers to the center frequency of its wavelet.
* Higher frequencies have more resolution, but do not penetrate as deep.

```python
def berlage_wavelet(t, f0, n=3, alpha=180, theta0=-np.pi/2, normalize=True):
    w = np.heaviside(t, 0.5) * t**n * np.exp(-alpha * t) * np.cos(2 * np.pi * f0 * t + theta0)
    if normalize:
        w /= np.max(np.abs(w))
    return w

def ricker_wavelet(t, f0, normalize=True):
    p = (np.pi * t * f0)**2
    w = (1 - 2 * p)*np.exp(-p)
    if normalize:
        w /= np.max(np.abs(w))
    return w

def do_wavelet_plot():
    f_0 = 1E8 # 100Mhz
    t = np.linspace(-3/f_0, 3/f_0, 1024)
    src1 = berlage_wavelet(t, f_0, n=4, alpha=6*f_0, theta0=0)
    src2 = ricker_wavelet(t, f_0)
    
    ref1 = np.cos(2 * np.pi * f_0 * t)
    ref2 = np.cos(2 * np.pi * f_0 * t)
    
    src1_tilde = np.fft.rfft(src1)
    src2_tilde = np.fft.rfft(src2)
    freqs = np.fft.rfftfreq(len(t), t[1]-t[0])
    
    plt.figure(figsize=(7.5,6), dpi=200)
    
    plt.subplot(2,2,1)
    plt.plot(t*1E9, src1)
    plt.plot(t*1E9, ref1, '--', alpha=0.5)
    plt.xlabel('Time (ns)')
    plt.ylabel('Berlage Wavelet')
    
    plt.subplot(2,2,2)
    plt.plot(freqs/1E6, np.abs(src1_tilde))
    plt.axvline(f_0/1E6, color='C1')
    plt.xlabel('Frequency (MHz)')
    plt.xlim([0, f_0/1E6 * 3])
    
    plt.subplot(2,2,3)
    plt.plot(t*1E9, src2)
    plt.plot(t*1E9, ref2, '--', alpha=0.5)
    plt.xlabel('Time (ns)')
    plt.ylabel('Ricker Wavelet')
    
    plt.subplot(2,2,4)
    plt.plot(freqs/1E6, np.abs(src2_tilde))
    plt.axvline(f_0/1E6, color='C1')
    plt.xlabel('Frequency (MHz)')
    plt.xlim([0, f_0/1E6 * 3])
    plt.tight_layout()
```

```python
do_wavelet_plot()
```
